```python
# Akıllı Gün Planlayıcı

while True:
    try:
        # Kullanıcıdan giriş alma
        uyku = float(input("Günlük uyku sürenizi (saat) giriniz: "))
        ders = float(input("Günlük ders sürenizi (saat) giriniz: "))
        serbest = float(input("Günlük serbest zamanınızı (saat) giriniz: "))

        # Negatif kontrolü
        if uyku < 0 or ders < 0 or serbest < 0:
            print("Hata: Negatif değer giremezsiniz!")
            continue

        # Toplam süre kontrolü
        toplam = uyku + ders + serbest
        if toplam > 24:
            print("Hata: Toplam süre 24 saati geçemez!")
            continue

        break  # Her şey doğruysa döngüden çık

    except:
        print("Hata: Lütfen sayısal bir değer giriniz!")

# Uyku durumu belirleme
if uyku < 6:
    uyku_durumu = "Yetersiz uyku"
elif 6 <= uyku <= 8:
    uyku_durumu = "Normal uyku"
else:
    uyku_durumu = "Fazla uyku"

# Çalışma durumu belirleme
if ders >= 6:
    calisma_durumu = "Yoğun çalışma günü"
else:
    calisma_durumu = "Normal çalışma günü"

# Genel değerlendirme
if uyku < 6 and ders >= 6:
    genel_durum = "Verimsiz gün"
else:
    genel_durum = "Dengeli gün"
```
